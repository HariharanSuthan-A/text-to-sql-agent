```python
from sqlite3 import connect

conn = connect("employees.db")   # persistent file
cursor = conn.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS employees (
        id INTEGER PRIMARY KEY,
        first_name TEXT NOT NULL,
        last_name TEXT NOT NULL,
        department TEXT NOT NULL,
        salary REAL NOT NULL,
        hire_date TEXT NOT NULL
    )
''')

employees = [
    (1, 'Alice', 'Johnson', 'Engineering', 95000.00, '2024-01-15'),
    (2, 'Bob', 'Smith', 'Sales', 76000.00, '2023-09-01'),
    (3, 'Carla', 'Lopez', 'Marketing', 82000.00, '2024-03-12'),
    (4, 'Daniel', 'Kim', 'Finance', 88000.00, '2022-11-28'),
    (5, 'Eve', 'Nguyen', 'Customer Support', 68000.00, '2024-04-02')
]

cursor.executemany('''
    INSERT OR REPLACE INTO employees
    VALUES (?, ?, ?, ?, ?, ?)
''', employees)

conn.commit()
conn.close()
```
